```python
class Car:

    def __init__(self, brand, model, color, price, year, fuel_type, engine, mileage):
        self.brand = brand
        self.model = model
        self.color = color
        self.price = price
        self.year = year
        self.fuel_type = fuel_type
        self.engine = engine
        self.mileage = mileage

    def show_details(self):
        print("Car Brand   :", self.brand)
        print("Car Model   :", self.model)
        print("Color       :", self.color)
        print("Price       :", self.price)
        print("Year        :", self.year)
        print("Fuel Type   :", self.fuel_type)
        print("Engine      :", self.engine)
        print("Mileage     :", self.mileage)

    def start(self):
        print(self.brand, self.model, "is started.")

    def stop(self):
        print(self.brand, self.model, "is stopped.")

    def drive(self):
        print(self.brand, self.model, "is running on the road.")
```

```python
class Car:

    def __init__(self, brand, model, color, price):
        self.brand = brand
        self.model = model
        self.color = color
        self.price = price

    def start(self):
        print(self.brand, "is started.")

    def stop(self):
        print(self.brand, "is stopped.")

    def show_details(self):
        print("Brand :", self.brand)
        print("Model :", self.model)
        print("Color :", self.color)
        print("Price :", self.price)
```

```python
class ElectricCar(Car):

    def charge_battery(self):
        print(self.brand, "battery is charging.")
```

```python
class BankAccount:

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

    def show_balance(self):
        print("Balance:", self.__balance)

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Amount deposited:", amount)
        else:
            print("Invalid amount")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Amount withdrawn:", amount)
        else:
            print("Insufficient balance")
```

```python
bank = BankAccount("amit", 60000)
```

```python
bank.name
```

```python
bank.show_balance()
```

```python
bank.deposit(50000)
```

```python
bank.show_balance()
```

```python
bank.withdraw(70000)
```

```python
bank.show_balance()
```

```python
class Car:

    def start(self):
        print("Car is starting")

    def stop(self):
        print("Car is stopping")

    def drive(self):
        print("Car is driving")


class ElectricCar(Car):

    def start(self):
        print("Electric car starts with a button")

    def drive(self):
        print("Electric car is driving silently")


class SportsCar(Car):

    def start(self):
        print("Sports car starts very quickly")

    def drive(self):
        print("Sports car is driving very fast")
```

```python
car1 = Car()
car2 = ElectricCar()
car3 = SportsCar()
```

```python
cars = [car1, car2, car3]
```

```python
for i in cars:
    i.start()
    i.drive()
    i.stop()
```
