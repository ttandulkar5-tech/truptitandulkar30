```python
num = 10

prime = True

for i in range(2, num):
    if num % i == 0:
        prime = False
        break

if prime:
    print("Prime")
else:
    print("Not Prime")
```

```python
text = "bananab"

visited = ""

for ch in text:
    if ch not in visited:
        print(ch, "=", text.count(ch))
        visited += ch
```

```python
ch = 65  

for i in range(1, 8):
    for j in range(i):
        print(chr(ch), end=" ")
        if chr(ch) == "Z":
            break
        ch += 1
    print()
```

```python
ch = 65       
row = 1  # 2 3 4

while row <= 7:   # T T T  
    col = 1   

    while col <= row:  # T T T F
        print(chr(ch), end=" ")
        if chr(ch)=="Z":
            break
        ch += 1  # 66 67 68 69 70
        col += 1  # 2 3 4

    print()
    row += 1
```

```python

row = 1

while row <= 7:
    col = 1

    while col <= row:
        print("*", end=" ")
        col += 1

    print()
    row += 1
```

```python
row = 1

while row <= 5:

    space = 5 - row
    while space > 0:
        print(" ", end="")
        space -= 1

    star = 1
    while star <= (2 * row - 1):
        print("*", end="")
        star += 1

    print()
    row += 1
```
