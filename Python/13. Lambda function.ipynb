```python
def str1(list1):
    lis=[]
    for i in list1:
        if type(i) == str:
            lis.append(i)
        elif type(i)== list or type(i)== set or type(i)== tuple:
            for j in i:
                if type(j) == str :
                    lis.append(j)
    return " ".join(lis)
```

```python
def count_up(n):
    for i in range(1, n + 1):
        yield i
```

```python
list(count_up(4))
```

```python
gen = count_up(5)
print(type(gen))  
```

```python
for val in gen:
    print(val, end=" ")   
```
