```python
from datetime import date, timedelta

date1 = date(2024, 1, 1)
date2 = date(2024, 12, 31)

# Find difference between dates
difference = date2 - date1
print(f"Days between dates: {difference.days}")  # Output: 364

# Check if date1 is before date2
print(date1 < date2)  # Output: True
```

```python
from datetime import time

# Create specific time
meeting_time = time(14, 30, 0)  # 2:30 PM
print(meeting_time)  # Output: 14:30:00

# Access time components
print(f"Hour: {meeting_time.hour}")
print(f"Minute: {meeting_time.minute}")
print(f"Second: {meeting_time.second}")
```

```python
from datetime import datetime

# Get current datetime
now = datetime.now()
print(now)  # Output: 2024-08-14 15:45:32.123456

# Get only time component
current_time = now.time()
print(current_time)  # Output: 15:45:32.123456

# Microseconds useful for precise measurements
print(f"Microseconds: {current_time.microsecond}")
```

```python
from datetime import datetime, timedelta

# Create datetime with specific time
start_time = datetime(2024, 8, 14, 9, 0, 0)  # 9:00 AM
print(f"Start: {start_time}")

# Add 2 hours 30 minutes
end_time = start_time + timedelta(hours=2, minutes=30)
print(f"End: {end_time}")

# Calculate duration
duration = end_time - start_time
print(f"Session duration: {duration}")  # Output: 2:30:00
```
